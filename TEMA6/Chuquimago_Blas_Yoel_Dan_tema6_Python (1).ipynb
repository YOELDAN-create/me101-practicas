```python

import numpy as np
import matplotlib.pyplot as plt

fila = np.linspace(0, 255, 5, dtype=np.uint8)

gris = np.tile(fila, (5, 1))

print("Imagen:")
print(gris)
print("Shape:", gris.shape)
print("Tipo de dato:", gris.dtype)
print("Media:", gris.mean())
```

```python

plt.imshow(gris, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen en escala de grises")
plt.axis("off")

plt.savefig("gris_tema6.png", bbox_inches="tight", pad_inches=0)
plt.show()
```

```python

recorte = gris[:, -2:]
print("Recorte de la mitad derecha:")
print(recorte)
```

```python
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = [255, 0, 0]
rgb[:, 2:4] = [0, 0, 255]

print("Dimensiones:", rgb.shape)
print("Pixel [0,0]:", rgb[0, 0])
print("Pixel [0,3]:", rgb[0, 3])

plt.imshow(rgb)
plt.title("Imagen RGB: rojo y azul")
plt.axis("off")
plt.show()
```

```python

R = rgb[:, :, 0]
G = rgb[:, :, 1]
B = rgb[:, :, 2]

print("Canal rojo:")
print(R)

print("Canal verde:")
print(G)

print("Canal azul:")
print(B)
```

```python

gris_promedio = np.mean(rgb, axis=2)

gris_luminosidad = (
    0.299 * rgb[:, :, 0]
    + 0.587 * rgb[:, :, 1]
    + 0.114 * rgb[:, :, 2]
)

print("Gris por promedio:")
print(gris_promedio)

print("Gris por luminosidad:")
print(gris_luminosidad)
```

```python
umbral = 50

binaria = np.where(gris_luminosidad > umbral, 255, 0).astype(np.uint8)

print("Imagen binarizada:")
print(binaria)

plt.imshow(binaria, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen binarizada")
plt.axis("off")
plt.show()
```
